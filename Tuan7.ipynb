Nhập vào một dòng gồm các số nguyên. Hãy loại bỏ phần tử trùng nhưng
giữ nguyên thứ tự xuất hiện đầu tiên, rồi in ra list kết quả.
Yêu cầu in ra đúng dạng list Python, ví dụ [1, 2, 3]


```python
n = list(map(int,input().split()))
lst = []
seen = set()
for i in n:
    if i not in lst:
        lst.append(i)
        seen.add(i)

print(lst)
```

Nhập một dòng số nguyên. In ra list tổng tích lũy (phần tử i là tổng từ đầu
tới vị trí i), giữ nguyên thứ tự

```python
n = list(map(int,input().split()))
lst = []
total = 0

for i in n:
    total += i
    lst.append(total)

print(lst)
```

Nhập một dòng gồm các cặp key:value cách nhau bởi dấu cách. Tạo
dictionary ánh xạ mỗi key -> list các value theo thứ tự xuất hiện. Nếu key
lặp lại, thêm vào list của key đó

```python
s = input()

result = {}

pairs = s.split()

for p in pairs:
    key, value = p.split(':')
    value = int(value)  # nếu value là số

    if key not in result:
        result[key] = []

    result[key].append(value)

print(result)
```

Nhập một list số nguyên, In ra dictionary {'positives': ..., 'negatives': ...,
'zeros': ...} (đếm số >0, <0, =0)


```python
def count_num(lst):
    result = {
        'positive': 0,
        'negative': 0,
        'zeros': 0,
    }

    for i in lst:
        if i > 0:
            result['positive'] += 1
        elif i < 0:
            result['negative'] += 1
        elif i == 0:
            result['zeros'] += 1
            
    return result

lst = list(map(int,input().split()))
print(count_num(lst))
```

dictionary mới bằng cách đảo ngược key và value của dictionary ban đầu. In
ra dictionary mới. 

```python
s = input().split()

d = {}

n = len(s)

for i in range(0,n,2):
    key = s[i]
    value = s[i+1]

    d[value] = key

print(d)
```

Nhập vào một list các số nguyên. Hãy tìm số xuất hiện nhiều lần nhất trong list.
Nếu có nhiều số cùng có số lần xuất hiện nhiều nhất, hãy in ra số nhỏ nhất
trong số đó.

```python
my_list = list(map(int,input().split()))

count = {}

for i in my_list:
    count[i] = count.get(i,0) + 1

max_count = 0
result = None

for i in my_list:
    if count[i] > max_count:
        max_count = count[i]
        result = i

print(result)
```

Viết chương trình nhập vào hai dictionary. Sau đó gộp lại thành một dictionary mới
trong đó nếu một key xuất hiện trong cả hai dict, giá trị sẽ được cộng lại, nếu chỉ có
trong một dict thì giữ nguyên. In ra dictionary mới theo thứ tự từ điển tăng dần của
key. 


```python
d1 = eval(input())
d2 = eval(input())
result = {}

for key, value in d1.items():
    result[key] = value

for key, value in d2.items():
    if key in result:
        result[key] += value
    else:
        result[key] = value

for key in sorted(result):
    print(key, ":", result[key])
```
