Cho một mảng số nguyên không âm. Gọi
mảng đó là "mảng đẹp" nếu mỗi phần tử
x trong mảng xuất hiện đúng x lần.
Kiểm tra xem mảng có phải là mảng đẹp
hay không. Nếu không hãy xóa đi y số (y
là nhỏ nhất có thể) để mảng còn lại thành
mảng đẹp.


```python
from collections import Counter

n = list(map(int,input().split()))

res = Counter(n)

xoa = 0

for i in range(len(n)):
    if res[i] == i:
        continue
    elif res[i] > i:
        xoa += res[i]-i
    else:
        xoa += res[i]

if xoa == 0:
    print("YES")
else:
    print("NO")
    print(xoa)
```

Tìm phần tử xuất hiện nhiều hơn n/2 lần
trong mảng (nếu có) với độ phức tạp thời
gian là O(n), độ phức tạp bộ nhớ là O(1),
- Boyer–Moore. Nếu không có phần tử
nào thỏa mãn thì không in gì

```python
a = list(map(int, input().split()))

candidate = None
count = 0

for x in a:
    if count == 0:
        candidate = x
        count = 1
    elif x == candidate:
        count += 1
    else:
        count -= 1

if a.count(candidate) > len(a) // 2:
    print(candidate)
else:
    print("Không có")

```

Viết hàm tìm kiếm nhị phân
trên một danh sách đã
được sắp xếp tăng dần.
Trả về chỉ số của phần tử
hoặc -1 nếu không tìm
thấy. mẫu hàm: def
binary_search(arr, left,
right, target):

```python
def binary_search(arr, left, right, target):
    while left <= right:
        mid = (left + right)//2

        if arr[mid] == target:
            return mid
        elif arr[mid] < target:
            left = mid + 1
        else:
            right = mid - 1

    return -1

arr = list(map(int, input().split()))
target = int(input())

print(binary_search(arr, 0, len(arr) - 1, target))
```

Cho danh sách arr và một số nguyên x, đếm xem x xuất hiện bao nhiêu lần trong danh sách. def count_occurrences(arr, x) - > int:

```python
def count_occurrences(arr, x) -> int:
    return arr.count(x)

arr = list(map(int,input().split()))
x = int(input())
print(count_occurrences(arr, x))
```

Viết hàm thực hiện thuật
toán bubble sort, Selection
Sort để sắp xếp danh sách
theo thứ tự tăng dần. Sắp
xếp nổi bọt (Bubble Sort); 


```python
def bubble_sort(arr):
    n = len(arr)

    for i in range(n):
        swap = False
        
        for j in range(0, n-i-1):
            if arr[j] > arr[j+1]:
                arr[j] , arr[j+1] = arr[j+1], arr[j]
                swap = True
        if not swap:
            break
    return arr
arr = list(map(int,input().split()))
print(bubble_sort(arr))
```

Viết hàm tìm phần tử có số
lần xuất hiện nhiều nhất
trong danh sách. Nếu có
nhiều phần tử cùng số lần,
trả về phần tử xuất hiện
trước.


```python
def most_frequent(arr):
    
    count = {}

    for x in arr:
        count[x] = count.get(x,0)+1

    max_count = 0
    result = None

    for x in arr:
        if count[x] > max_count:
            max_count = count[x]
            result = x

    return f"{result} xuất hiện nhiều nhất, sớm nhất và {max_count} lần"

arr = list(map(int,input().split()))
print(most_frequent(arr))
```

Cho danh sách các số
nguyên và một số X, đếm
tất cả các cặp (a, b) sao
cho a + b = X. c


```python
def count_pairs(arr, X):
    count = 0
    freq = {}

    for num in arr:
        target = X - num

        if target in freq:
            count += freq[target]

        freq[num] = freq.get(num, 0) + 1

    return count


arr = list(map(int, input().split()))
X = int(input())

print(count_pairs(arr, X))
```

Bạn được cung cấp một

lồng nhau, vd [[1, 2, 3], [4,

Cho một danh sách các số
nguyên nums, hãy tìm độ
dài của dãy con tăng
nghiêm ngặt dài nhất
(không nhất thiết phải liền kề)

```python
arr = eval(input())

n = len(arr)

dp = [1]*n

for i in range(n):
    for j in range(i):
        if arr[j] < arr[i]:
            dp[i] = max(dp[i], dp[j]+1)

print(max(dp))
```
