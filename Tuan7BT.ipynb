```python
def min_and_position(lst):
    if lst:
        Min_value = lst[0]
        min_pos = 1

        for i, j in enumerate(lst,1):
            if j < Min_value:
                Min_value = lst[i]
                min_pos = j
        return Min_value, min_pos
    return None

assert min_and_position([10,0,0,3,4,6,6]) == (0,2)
print("Pass all test!")



```

Tìm giá trị trung bình của các sô nguyên dương trong lst nếu không trả về "invalid"


```python
def average_num(lst):
    if lst:
        t = 0
        d = 0

        for x in lst:
            if x > 0:
                t += x
                d += 1
        if not d:
            return "Invalid"
       
        return round(t/d,2)
    return "Invalid"

assert average_num([1,2,3,4]) == 2.5
assert average_num([-1,-2,-1]) == "Invalid"
assert average_num([]) == "Invalid"
print("OK!")

        

            
```

```python
def doi_xung(lst):
    if lst:
       l = len(lst)
       for i in range(l//2):
            if lst[i] != lst[l-1-i]:
                return False
       return True
    return None

def kt_doi_xung(lst1):
    if lst1:
      lst1 = [1,2,3,2,1]
      lst2 = list(lst1)
      lst1.reverse()
      return lst1 == lst2
    return None
    

assert doi_xung([1,2,3,2,1]) == True
assert doi_xung([1,2,3,3,1]) != True
assert doi_xung([]) == None


print("Accepted!!")
```

```python
def move_0_to_first(lst):
    if lst:
        lst_zero = []
        lst_non_zero = []
        for x in lst:
            if x == 0:
                lst_zero.append(0)
            else:
                lst_non_zero.append(x)
        return lst_zero + lst_non_zero
    return None


assert move_0_to_first([3,4,6,7,1,9,0,0,23,0,0,34,47,69]) == [0,0,0,0,3,4,6,7,1,9,23,34,47,69]
assert move_0_to_first([0,4,0,0]) == [0,0,0,4]
print("Accepted!!")

```

```python
def Unique_array(lst):
    if lst:
        ans = []
        for i in lst:
            if i not in ans:
                ans.append(i)
        return ans
    return None

assert Unique_array([1,1,1,2,2,2,3,4,4,4]) == [1,2,3,4]
print("True!")
```
